Task 1 — Text Cleaning System
Objective

Build a preprocessing pipeline that cleans raw text.

```python
import re
import string
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

text = "Hello!!! I am learning NLP @ CVR College :)"

# convert to lowercase
text = text.lower()

# remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

# remove numbers/special characters using regex
text = re.sub(r'[^a-zA-Z\s]', '', text)

# remove stopwords
stop_words = set(stopwords.words('english'))

words = text.split()

filtered_words = [word for word in words if word not in stop_words]

clean_text = " ".join(filtered_words)

print("Cleaned Text:")
print(clean_text)
```

Task 2 — Tokenization Analyzer
Objective

Convert paragraphs into:

sentences
words
tokens

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt_tab')

paragraph = """
Natural Language Processing is amazing.
It helps computers understand human language.
Tokenization is an important NLP technique.
"""

# sentence tokenization
sentences = sent_tokenize(paragraph)

# word tokenization
words = word_tokenize(paragraph)

# unique words
unique_words = set(words)

print("Sentences:")
print(sentences)

print("\nWords:")
print(words)

print("\nTotal Sentences:", len(sentences))
print("Total Words:", len(words))
print("Unique Words:", len(unique_words))
```

Task 3 — Stopword Removal Tool
Objective

Remove unnecessary words from text.

```python
from nltk.corpus import stopwords

text = "This is an amazing NLP session for the students"

stop_words = set(stopwords.words('english'))

words = text.split()

filtered = [word for word in words if word.lower() not in stop_words]

result = " ".join(filtered)

print("After Stopword Removal:")
print(result)
```

Task 4 — Stemming vs Lemmatization Comparison
Objective

Compare stemming and lemmatization.

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

text = "running studies caring"

words = text.split()

# stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in words]

# lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word, pos='v')
    for word in words
]

print("Stemmed Words:")
print(stemmed_words)

print("\nLemmatized Words:")
print(lemmatized_words)
```

Task 5 — Bag of Words Implementation
Objective

Convert text into numerical vectors.

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

documents = [
    "NLP is fun",
    "I love NLP",
    "NLP is powerful"
]

# create BoW model
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

# vocabulary
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)

# document vectors
df = pd.DataFrame(
    X.toarray(),
    columns=vocabulary
)

print("\nDocument Vectors:")
print(df)
```

Task 6 — TF-IDF Important Word Detection
Objective

Identify important words in documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

documents = [
    "NLP is amazing and useful",
    "Machine learning uses NLP",
    "Deep learning and NLP are related"
]

# TF-IDF vectorizer
tfidf = TfidfVectorizer()

X = tfidf.fit_transform(documents)

# feature names
features = tfidf.get_feature_names_out()

# convert to dataframe
df = pd.DataFrame(
    X.toarray(),
    columns=features
)

print("TF-IDF Matrix:")
print(df)
```

```python
# =========================
# IMPORT LIBRARIES
# =========================

import re
import string
import pandas as pd

from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

import nltk

# =========================
# DOWNLOAD NLTK DATA
# =========================

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# =========================
# USER INPUT
# =========================

text = input("Enter your paragraph:\n")

# =========================
# TASK 1 - TEXT CLEANING
# =========================

print("\n========== TEXT CLEANING ==========")

# lowercase conversion
clean_text = text.lower()

# remove punctuation
clean_text = clean_text.translate(
    str.maketrans('', '', string.punctuation)
)

# regex cleaning
clean_text = re.sub(r'[^a-zA-Z\s]', '', clean_text)

# stopword removal
stop_words = set(stopwords.words('english'))

words = clean_text.split()

filtered_words = [
    word for word in words
    if word not in stop_words
]

clean_text = " ".join(filtered_words)

print("Cleaned Text:")
print(clean_text)

# =========================
# TASK 2 - TOKENIZATION
# =========================

print("\n========== TOKENIZATION ==========")

sentences = sent_tokenize(text)

tokens = word_tokenize(clean_text)

unique_words = set(tokens)

print("Sentences:")
print(sentences)

print("\nTokens:")
print(tokens)

print("\nTotal Sentences:", len(sentences))
print("Total Words:", len(tokens))
print("Unique Words:", len(unique_words))

# =========================
# TASK 3 - STOPWORD REMOVAL
# =========================

print("\n========== STOPWORD REMOVAL ==========")

print("After Removing Stopwords:")
print(clean_text)

# =========================
# TASK 4 - STEMMING VS LEMMATIZATION
# =========================

print("\n========== STEMMING VS LEMMATIZATION ==========")

stemmer = PorterStemmer()

lemmatizer = WordNetLemmatizer()

stemmed_words = [
    stemmer.stem(word)
    for word in tokens
]

lemmatized_words = [
    lemmatizer.lemmatize(word, pos='v')
    for word in tokens
]

print("Stemmed Words:")
print(stemmed_words)

print("\nLemmatized Words:")
print(lemmatized_words)

# =========================
# TASK 5 - BAG OF WORDS
# =========================

print("\n========== BAG OF WORDS ==========")

documents = [clean_text]

bow = CountVectorizer()

bow_matrix = bow.fit_transform(documents)

vocabulary = bow.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vocabulary
)

print("\nBoW Vector:")
print(bow_df)

# =========================
# TASK 6 - TF-IDF
# =========================

print("\n========== TF-IDF ==========")

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(documents)

tfidf_words = tfidf.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_words
)

print("TF-IDF Matrix:")
print(tfidf_df)

# =========================
# IMPORTANT WORDS
# =========================

print("\n========== IMPORTANT WORDS ==========")

scores = zip(
    tfidf_words,
    tfidf_matrix.toarray()[0]
)

sorted_scores = sorted(
    scores,
    key=lambda x: x[1],
    reverse=True
)

for word, score in sorted_scores:
    print(f"{word} --> {score:.4f}")
```
