**STEMMING**-reduce words to root form

-> remove prefixes and suffixes

EX: "likes"|"liked"|"likely"|"liking"->"like"

```python
from nltk.stem import PorterStemmer #basic stemmer
porter_stemmer=PorterStemmer()
words=['Running','jumps','liked','reading','playing','happily','bookigs']
stemmed_words=[porter_stemmer.stem(word) for word in words]
print("Original words:",words)
print("stemmed_words",stemmed_words)
```

```python
#Snowball stemmer : is an enhanced version of the Porter stemmer, Supports multiple languages.
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer(language='english')
words_to_stem=["running","jumped","Happily","Quickly","foxes"]
stemmed_words=[snowball_stemmer.stem(word) for word in words_to_stem]
print("Original words:",words_to_stem)
print("stemmed word:",stemmed_words)
```

**TEXT REPRESENTATION TECHNIQUES**

manchine donot understand taxt so we convert it into numbers

1.**Bag of Words (BOW)**

  represents text as word frequency vectors.(ignores order and grammar)

  use: text classification,sentiment analysis,clustering


**step1:preprocessing the text**

clean the text (convert to lower case , remove spl characters)


```python
import re
dataset=[]
text=[
    "I Love coding",
    "I love Python#$"
    "Python coding is   fun"
]
for sentence in text:
  sentence=sentence.lower()
  sentence=re.sub(r'\W',' ',sentence)#remove symbols
  sentence=re.sub(r'\s+',' ',sentence)#remove extra spaces
  dataset.append(sentence)
print(dataset)
```

**step 2:Counting word Frequencies**

count how many times each word appears

```python
from collections import defaultdict
word2count=defaultdict(int)
for sentence in dataset:
  for word in sentence.split():
    word2count[word]+=1
print(word2count)
```

**step 3: Selecting Most Frequent Words**

let's take top 3 words

```python
import heapq

freq_words=heapq.nlargest(3,word2count,key=word2count.get)
print(freq_words)
```

**step 4: Build Bag of words Matrix**

rows=sentences

columns=frequent words

value=1(present) or 0(absent)

```python
x=[]
for sentence in dataset:
  vector=[]
  words=sentence.split()
  for word in freq_words:
    if word in freq_words:
      vector.append(1)
    else:
      vector.append(0)
  x.append(vector)
print(x)
```

**N-Grams

capture word sequences

splits text into sequences of N consecutive words to capture the context and word order of ml models

Ex:

1 word-> unigram ['i','love','coding']

2 words->bigram ['i love','love coding']

3 words->Trigram ['i love coding']

```python
#step 4: Preprocessing the Text
import re
text=[
    "I love coding",
    "I love Python",
    "Python coding is fun"
]
dataset=[]
for sentence in text:
  sentence=sentence.lower()
  sentence=re.sub(r'\W',' ',sentence)
  sentence=re.sub(r'\s+',' ',sentence)
  dataset.append(sentence)
print(dataset)
```

```python
#Step 2: create N-grams
def generate_ngrams(sentence,n):
  words=sentence.split()
  ngrams=[]
  for i in range(len(words)-n+1):
    ngrams.append(" ".join(words[i:i+n]))
  return ngrams
```

```python
#Ex : Bigrams(N=2)
for sentence in dataset:
  bigrams=generate_ngrams(sentence,2)
print(bigrams)

#Ex : Trigrams(N=3)
for sentence in dataset:
  trigrams=generate_ngrams(sentence,3)
print(trigrams)
```

```python
#step 3: Count N-Gram Frequencies

from collections import defaultdict
ngram2count=defaultdict(int)

for sentence in dataset:
  for ngram in generate_ngrams(sentence,2):
    ngram2count[ngram]+=1
print(ngram2count)
```

```python
#step 4: Convert into feature vector (Like Bow):
#select top N-Grams:
import heapq
freq_ngrams=heapq.nlargest(2,ngram2count,key=ngram2count.get)
print(freq_ngrams)
```

```python
#Build Matrix
x=[]
for sentence in dataset:
  vector=[]
  sentence_ngrams=generate_ngrams(sentence,2)

  for gram in freq_ngrams:
    if gram in sentence_ngrams:
      vector.append(1)
    else:
      vector.append(0)
  x.append(vector)
print(x)
```

###**TF-IDF(Term Frequncy-Inverse Document Frequency)**

->gives importance to rare words. used to measur how important as word is in the document

why TF-IDF?

BOW treats all words equally bcz of which words may occur frequently, tf-idf gives weights words based on importance

**step 1: Term Frequency(TF)**

*   how often a word appears in a sentence
*   formula:
     Tf=count of word in sentence/total words in sentnce




**step 2:Inverse doc frequency (IDF)**

how rare a word is acrross all sentences


IDF=log()

**step 3:TF-IF Calculation**

TF-IDF=TF*idf

```python
!pip install gensim

from gensim.models import Word2Vec

sentences = [
    ["i", "love", "coding"],
    ["i", "love", "python"],
    ["python", "is", "powerful"]
]

# Train CBOW model (sg = 0)
cbow_model = Word2Vec(
    sentences,
    min_count=1,
    vector_size=5,   # size of word vectors
    window=2,        # context window
    sg=0             # 0 = CBOW
)

# Get word vector
print("Vector for coding:")
print(cbow_model.wv["coding"])

# Find similar words
print("Similar to coding:")
print(cbow_model.wv.most_similar("coding"))



sentence="Elon visited India"
words=nltk.word_tokenize(sentence)
pos_tags=nltk.pos_tag(words)
ner_tree=nltk.ne_chunk(pos_tags)
print(ner_tree)
```
